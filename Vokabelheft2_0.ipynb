```python
# Vokabelbuch als Python-Programm

python_grundlagen = {
    "arithmetische operatoren": "Arithmetische Operatoren werden verwendet, um mathematische Berechnungen mit Zahlen durchzuführen, z. B. +, -, *, /, //, %, **.",
    "function": "Ein wiederverwendbarer Codeblock, der eine bestimmte Aufgabe ausführt. Funktionen können Parameter erhalten und Werte zurückgeben.\nBeispiel:\ndef addiere(a, b):\n    return a + b",
    "for": "Die for-Schleife wird verwendet, um einen Codeblock wiederholt auszuführen – für jedes Element in einer Sequenz (z. B. Liste, Tupel, String).\nBeispiel:\nfor zahl in [1, 2, 3]:\n    print(zahl)",
    "int": "int steht für ganze Zahlen – also ohne Komma. Zum Beispiel: 10, -3 oder 0. Du kannst mit ihnen rechnen (addieren, subtrahieren, multiplizieren etc.).",
    "float": "float steht für Gleitkommazahlen – Zahlen mit Nachkommastellen, wie 3.14 oder -0.5. Auch mit ihnen kannst du rechnen wie mit int.",
    "str": "str steht für String, also Text. Zeichenketten wie 'Hallo' oder \"123\". Sie können verknüpft, durchsucht oder in Teile geschnitten werden.",
    "bool": "bool steht für Wahrheitswerte: True (wahr) oder False (falsch). Diese nutzt man z. B. in Bedingungen oder Vergleichen.",
    "list": "Eine Liste ist eine geordnete Sammlung von Elementen. Sie ist veränderbar. Beispiel: [1, 2, 3] oder ['Hund', 'Katze'].",
    "tuple": "Ein Tupel ist wie eine Liste, aber unveränderbar (nicht editierbar). Beispiel: (1, 2) oder ('a', 'b').",
    "dict": "Ein Dictionary besteht aus Schlüssel-Wert-Paaren. Beispiel: {'Name': 'Anna', 'Alter': 20}. Man greift über den Schlüssel zu.",
    "set": "Ein Set ist eine ungeordnete Sammlung von eindeutigen Elementen. Duplikate werden automatisch entfernt. Beispiel: {1, 2, 3}.",
    "data type": "Ein Datentyp beschreibt, welche Art von Wert eine Variable hat – z. B. int (Zahl), str (Text), bool (True/False), float (Kommazahl).",
    "variable": "Eine Variable ist ein Name, unter dem ein Wert gespeichert ist. Beispiel: name = \"Lisa\" speichert den Text 'Lisa' in der Variable name.",
    "operator": "Ein Symbol, das etwas mit Werten macht – z. B. + addiert, * multipliziert. Weitere: -, /, %, ==, !=.",
    "expression": "Ein Ausdruck ist eine Kombination aus Werten, Variablen und Operatoren, die zu einem Ergebnis führen. Beispiel: (3 + 2) * 4 ergibt 20.",
    "statement": "Eine Anweisung führt eine Aktion aus. Beispiel: x = 5 ist eine Zuweisung. print('Hi') ist eine Ausgabe.",
    "class": "Eine Klasse ist eine Vorlage für Objekte. Sie definiert Eigenschaften und Methoden. Beispiel:\nclass Hund:\n    def __init__(self, name):\n        self.name = name",
    "objekt": "Ein Objekt ist eine Instanz einer Klasse. Beispiel: mein_hund = Hund(\"Bello\") erstellt ein Hund-Objekt.",
    "method": "Eine Methode ist eine Funktion, die zu einem Objekt gehört. Sie kann auf Daten des Objekts zugreifen.",
    "module": "Ein Modul ist eine Datei mit Python-Code. Es kann importiert werden, um Funktionen wiederzuverwenden. Beispiel: import math",
    "package": "Ein Paket ist eine Sammlung von Modulen, die zusammengehören. Beispiel: Das Paket 'numpy' enthält viele Mathe-Funktionen.",
    "library": "Eine Bibliothek ist eine Sammlung von Modulen und Paketen. Sie bieten viele Funktionen, z. B. zum Rechnen, Web-Entwicklung oder Datenanalyse.",
    "loop": "Eine Schleife wiederholt Code. Mit einer for- oder while-Schleife kannst du z. B. mehrmals etwas ausgeben.",
    "conditional": "Bedingte Anweisungen wie if, elif, else führen nur dann Code aus, wenn eine Bedingung erfüllt ist.",
    "debugging": "Debugging ist das Suchen und Beheben von Fehlern im Code.",
    "exception": "Ein Fehler, der beim Ausführen des Codes auftritt. Beispiel: ValueError, wenn man statt einer Zahl Text eingibt.",
    "+": "Addition: 5 + 3 = 8",
    "-": "Subtraktion: 5 - 2 = 3",
    "*": "Multiplikation: 3 * 2 = 6",
    "/": "Division: 5 / 2 = 2.5",
    "//": "Ganzzahlige Division: 5 // 2 = 2 (ohne Nachkommastelle)",
    "%": "Modulo: 5 % 2 = 1 (Rest der Division)",
    "**": "Potenzierung: 2 ** 3 = 8 (2 hoch 3)",
    "!=": "Ungleich: 5 != 3 ergibt True",
    "==": "Gleich: 5 == 5 ergibt True",
    ">": "Größer als: 6 > 3 ergibt True",
    "<": "Kleiner als: 2 < 5 ergibt True",
    ">=": "Größer gleich: 3 >= 3 ergibt True",
    "<=": "Kleiner gleich: 2 <= 4 ergibt True",
    "comment": "Kommentare erklären Code und werden vom Interpreter ignoriert. Sie beginnen mit #.",
    "keyword": "Reservierte Wörter in Python wie if, else, def. Sie dürfen nicht als Variablennamen verwendet werden.",
    "input": "input() holt Eingaben vom Benutzer. Beispiel: name = input('Wie heißt du?')",
    "interpreter": "Das Programm, das deinen Python-Code Zeile für Zeile ausführt.",
    "IDE": "Eine Entwicklungsumgebung (IDE) hilft beim Schreiben von Code – z. B. mit Syntaxhervorhebung oder Debugging.",
    "if": "Führt Code aus, wenn eine Bedingung wahr ist. Beispiel:\nif x > 0:\n    print('Positiv')",
    "else": "Führt Code aus, wenn die vorherige Bedingung nicht erfüllt war.",
    "elif": "Zusätzliche Bedingung, wenn das erste if nicht zutrifft.",
    "while": "Solange eine Bedingung wahr ist, wird der Codeblock ausgeführt.",
    "print": "Gibt Text oder Werte auf dem Bildschirm aus.",
    "true": "Der boolesche Wert für 'wahr'. Wird in Bedingungen verwendet.",
    "def": "Wird verwendet, um eine Funktion zu definieren. Beispiel:\ndef sag_hallo():\n    print('Hallo!')"
}



vokabular = python_grundlagen
def suchfunktion():
    begriff = input("\nWonach willst du suchen? ").strip().lower()
    if begriff in vokabular:
        print(f"\n{begriff}:\n{vokabular[begriff]}")
    else:
        print("Dieser Begriff wurde leider nicht gefunden.")

def alle_anzeigen():
    for begriff in sorted(vokabular):
        print(f"\n{begriff}:\n{vokabular[begriff]}")

def menü():
    print("📘 Willkommen im Python-Vokabelbuch 📘")
    while True:
        print("\nWas möchtest du tun?")
        print("1 - Nach einem Begriff suchen")
        print("2 - Alle Begriffe anzeigen")
        print("0 - Beenden")
        wahl = input("Deine Wahl: ").strip()

        if wahl == "1":
            suchfunktion()
        elif wahl == "2":
            alle_anzeigen()
        elif wahl == "0":
            print("Tschüss! 👋")
            break
        else:
            print("Ungültige Eingabe. Bitte versuche es erneut.")

# Programmstart
if __name__ == "__main__":
    menü()
```
