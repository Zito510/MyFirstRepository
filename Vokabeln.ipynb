Funktionsweise:

Die Funktion erklaerung_abrufen() dient dazu, die Erklärung einer Vokabel aus einem Vokabular abzurufen und anzuzeigen. Sie überprüft die Existenz der Vokabel, ruft die Erklärung ab, formatiert sie und gibt sie auf der Konsole aus. Falls die Vokabel nicht gefunden wird, wird eine Fehlermeldung ausgegeben.

```python
def erklaerung_abrufen(vokabel):
    """Diese Funktion gibt die Erklärung einer Vokabel aus,
    falls sie im Vokabular vorhanden ist."""

    # Überprüfen, ob die Vokabel im Vokabular vorhanden ist
    if vokabel in vokabular:
        # Die Erklärung der Vokabel aus dem Vokabular abrufen
        erklaerung = vokabular[vokabel]

        # Escape-Sequenzen für Zeilenumbrüche durch tatsächliche Zeilenumbrüche ersetzen
        erklaerung = erklaerung.replace("\\n", "\n")

        # Die Erklärung auf der Konsole ausgeben
        print(erklaerung)
    # Falls die Vokabel nicht gefunden wurde
    else:
        # Eine Meldung ausgeben, dass die Vokabel nicht gefunden wurde
        print("Vokabel nicht gefunden.")
```

Funktionsweise:

Die Funktion rueckwaertssuche() ermöglicht es, im Vokabular nach Vokabeln zu suchen, deren Erklärungen einen bestimmten Suchbegriff enthalten. Sie ist nützlich, um schnell Vokabeln zu finden, die sich auf ein bestimmtes Thema oder Konzept beziehen.


```python
def rueckwaertssuche(suchbegriff):
    """Diese Funktion sucht nach Vokabeln, deren Erklärungen einen bestimmten Suchbegriff enthalten."""

    # Ein leeres Dictionary erstellen, um die Ergebnisse zu speichern
    ergebnisse = {}

    # Durch alle Vokabeln und Erklärungen im Vokabular iterieren
    for vokabel, erklaerung in vokabular.items():
        # Escape-Sequenzen für Zeilenumbrüche durch tatsächliche Zeilenumbrüche ersetzen
        erklaerung = erklaerung.replace("\\n", "\n")
        # Überprüfen, ob der Suchbegriff in der Erklärung vorkommt (ohne Berücksichtigung der Groß-/Kleinschreibung)
        if re.search(suchbegriff, erklaerung, re.IGNORECASE):
            # Wenn der Suchbegriff gefunden wurde, die Vokabel und die Erklärung zum Ergebnis-Dictionary hinzufügen
            ergebnisse[vokabel] = erklaerung
    # Das Ergebnis-Dictionary zurückgeben
    return ergebnisse
```

Funktionsweise:

Die while-Schleife ermöglicht es dem Benutzer, wiederholt Vokabeln abzufragen oder nach Vokabeln zu suchen, die einen bestimmten Suchbegriff in ihrer Erklärung enthalten. Die Schleife läuft so lange, bis der Benutzer sie durch Eingabe von 'q' beendet.

Diese Funktion nimmt einen Suchbegriff als Argument entgegen und gibt eine Liste aller Vokabeln zurück, deren Erklärungen den Suchbegriff enthalten. Sie verwendet re.search mit dem Flag re.IGNORECASE, um eine Suche ohne Berücksichtigung der Groß- und Kleinschreibung durchzuführen.

```python
while True:
    # Eingabeaufforderung für den Benutzer
    eingabe = input("\nGib eine Vokabel oder einen Suchbegriff ein (oder 'q' zum Beenden): ")
    # Überprüfen, ob der Benutzer die Schleife beenden möchte
    if eingabe.lower() == 'q':
        break  # Schleife beenden, wenn der Benutzer 'q' eingibt

    # Überprüfen, ob die Eingabe eine Vokabel ist
    if eingabe in vokabular:
        # Erklärung der Vokabel abrufen und ausgeben
        erklaerung_abrufen(eingabe)
    # Wenn die Eingabe kein Vokabel ist, wird eine Rückwärtssuche durchgeführt
    else:
        # Rückwärtssuche nach dem Suchbegriff
        ergebnisse = rueckwaertssuche(eingabe)
        # Wenn Ergebnisse gefunden wurden
        if ergebnisse:
            # Überschrift für die Ergebnisse ausgeben
            print("Gefundene Vokabeln:")
            # Ergebnisse durchgehen und ausgeben
            for vokabel, erklaerung in ergebnisse.items():
                print(f"{vokabel}: {erklaerung}")  # Vokabel und Erklärung ausgeben
                print()  # Leere Zeile für bessere Lesbarkeit
        # Wenn keine Ergebnisse gefunden wurden
        else:
            # Meldung ausgeben, dass keine Vokabeln gefunden wurden
            print("Keine Vokabeln gefunden, die den Suchbegriff enthalten.")
```
