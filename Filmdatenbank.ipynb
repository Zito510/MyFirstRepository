```python
# Filmklasse erstellen
class Film:
    def __init__(self, titel, regisseur, jahr, genre):
        self.titel = titel
        self.regisseur = regisseur
        self.jahr = jahr
        self.genre = genre
        self.bewertung = None  # Bewertung wird später gesetzt

    def info(self):
        info_text = f"Titel: {self.titel}\nRegisseur: {self.regisseur}\nJahr: {self.jahr}\nGenre: {self.genre}"
        if self.bewertung is not None:
            info_text += f"\nBewertung: {self.bewertung}/10"
        return info_text

    def bewerten(self, bewertung):
        if 0 <= bewertung <= 10:
            self.bewertung = bewertung
        else:
            raise ValueError("Bewertung muss zwischen 0 und 10 liegen.")


# Filme anlegen
matrix = Film("Matrix", "Lana Wachowski & Lilly Wachowski", 1999, "Science-Fiction")

star_wars_1 = Film("Star Wars: Episode I - Die dunkle Bedrohung", "George Lucas", 1999, "Science-Fiction/Fantasy")
star_wars_2 = Film("Star Wars: Episode II - Angriff der Klonkrieger", "George Lucas", 2002, "Science-Fiction/Fantasy")
star_wars_3 = Film("Star Wars: Episode III - Die Rache der Sith", "George Lucas", 2005, "Science-Fiction/Fantasy")
star_wars_4 = Film("Star Wars: Episode IV - Eine neue Hoffnung", "George Lucas", 1977, "Science-Fiction/Fantasy")
star_wars_5 = Film("Star Wars: Episode V - Das Imperium schlägt zurück", "Irvin Kershner", 1980, "Science-Fiction/Fantasy")
star_wars_6 = Film("Star Wars: Episode VI - Die Rückkehr der Jedi-Ritter", "Richard Marquand", 1983, "Science-Fiction/Fantasy")


# Beispiele der Anwendung
print(matrix.info())
print("\n")

# Filme bewerten
matrix.bewerten(9)
star_wars_4.bewerten(10)
star_wars_5.bewerten(10)

# Info nach Bewertung anzeigen
print(matrix.info())
print("\n")
print(star_wars_4.info())
print("\n")
print(star_wars_5.info())

# Alle Filme in einer Liste speichern und zusammen anzeigen
filme = [matrix, star_wars_1, star_wars_2, star_wars_3, star_wars_4, star_wars_5, star_wars_6]

print("\nAlle Filme:")
for film in filme:
    print("\n" + film.info())
```
